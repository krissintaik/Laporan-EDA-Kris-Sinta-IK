```python
# 1. Hubungkan Google Drive & Import Library
from google.colab import drive
drive.mount('/content/drive')

import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

# 2. Baca Dataset dari Google Drive
file_path = '/content/drive/MyDrive/Tugas_AMIKOM/FIK Olah SIPD - LPJ Fungsional - Januari s d Oktober 2026.xlsx'
df = pd.read_excel(file_path, sheet_name='LPJ Administratif 2026')

# 3. Tampilkan Dimensi & 5 Baris Pertama
print(f"Dimensi Dataset: {df.shape[0]} baris, {df.shape[1]} kolom")
display(df.head())

```

```python
# 1. Cek Tipe Data Awal
print("=== TIPE DATA AWAL ===")
print(df.dtypes)

# 2. Normalisasi Label Bulan & Urutan Kronologis
df['Bulan'] = df['Bulan'].astype(str).str.strip()
urutan_bulan = [
    'Januari', 'Februari', 'Maret', 'April', 'Mei',
    'Juni', 'Juli', 'Agustus', 'September', 'Oktober'
]
df['Bulan'] = pd.Categorical(df['Bulan'], categories=urutan_bulan, ordered=True)

# 3. Fungsi Konversi Format Rupiah Indonesia ke Float
def rupiah_to_float(val):
    if pd.isna(val):
        return np.nan
    if isinstance(val, (int, float, np.number)):
        return float(val)
    cleaned = str(val).replace('Rp', '').replace('.', '').replace(',', '.').strip()
    return float(cleaned)

kolom_rupiah = ['Jumlah Anggaran', 'SPJ - LS Gaji', 'SPJ - LS Barang & Jasa',
                'SPJ - UP/GU/TU', 'Jumlah SPJ (LS + UP/GU/TU)']
for col in kolom_rupiah:
    df[col] = df[col].map(rupiah_to_float)

# 4. Validasi Rekonsiliasi Komponen SPJ
selisih = (df['Jumlah SPJ (LS + UP/GU/TU)']
           - (df['SPJ - LS Gaji'] + df['SPJ - LS Barang & Jasa'] + df['SPJ - UP/GU/TU'])).abs().max()
print(f"\nMaksimum selisih rekonsiliasi SPJ: Rp{selisih:,.2f}")
print("\n=== TIPE DATA SETELAH PEMBERSIHAN ===")
print(df.dtypes)
```

```python
# Pengecekan Missing Values dan Nilai Nol
missing_per_col = df.isna().sum()
total_missing = missing_per_col.sum()
jumlah_nol_spj = (df['Jumlah SPJ (LS + UP/GU/TU)'] == 0).sum()

print("=== HASIL PEMERIKSAAN MISSING VALUES ===")
print(missing_per_col)
print("-" * 45)
print(f"Total sel kosong (NaN)   : {total_missing}")
print(f"Baris realisasi Rp0,00   : {jumlah_nol_spj} baris ({jumlah_nol_spj/len(df)*100:.2f}%)")
```

```python
from scipy.stats import skew, kurtosis

# 1. Hitung Pagu Unik & Agregasi Bulanan
kunci_rekening = ['Sub Unit', 'Program', 'Kegiatan', 'Sub Kegiatan', 'Kode Rekening', 'Uraian']
total_pagu = df[kunci_rekening + ['Jumlah Anggaran']].drop_duplicates()['Jumlah Anggaran'].sum()

bulanan = df.groupby('Bulan', observed=True).agg(
    Realisasi=('Jumlah SPJ (LS + UP/GU/TU)', 'sum')
).reset_index()
bulanan['Kontribusi_Pagu_%'] = bulanan['Realisasi'] / total_pagu * 100
bulanan['Kumulatif'] = bulanan['Realisasi'].cumsum()
bulanan['Serapan_Kumulatif_%'] = bulanan['Kumulatif'] / total_pagu * 100

print(f"Total Pagu Unik      : Rp{total_pagu:,.2f}")
print(f"Total Realisasi      : Rp{bulanan['Realisasi'].sum():,.2f}")
print(f"Sisa Pagu Anggaran   : Rp{total_pagu - bulanan['Realisasi'].sum():,.2f}\n")
display(bulanan)

# 2. Statistik Distribusi, Skewness, Kurtosis & IQR
spj = df['Jumlah SPJ (LS + UP/GU/TU)']
q1, med, q3 = np.percentile(spj, [25, 50, 75])
iqr = q3 - q1
batas_atas = q3 + 1.5 * iqr

print(f"\nQ1: Rp{q1:,.2f} | Median: Rp{med:,.2f} | Q3: Rp{q3:,.2f} | IQR: Rp{iqr:,.2f}")
print(f"Batas Atas IQR (Q3 + 1.5*IQR): Rp{batas_atas:,.2f}")
print(f"Jumlah Kandidat Pencilan     : {(spj > batas_atas).sum()} observasi")
print(f"Skewness: {skew(spj):.2f} | Excess Kurtosis: {kurtosis(spj):.2f}")
```

```python
# Visualisasi EDA Lengkap (4 Grafik Utama)
fig, axes = plt.subplots(2, 2, figsize=(15, 10))
plt.subplots_adjust(hspace=0.38, wspace=0.25)

# 1. Grafik Batang Realisasi Bulanan
axes[0, 0].bar(bulanan['Bulan'], bulanan['Realisasi'] / 1e9, color='#1F4E79')
axes[0, 0].set_title('Realisasi Belanja Bulanan Januari–Oktober 2026', fontweight='bold')
axes[0, 0].set_ylabel('Miliar Rupiah')
axes[0, 0].tick_params(axis='x', rotation=35)
axes[0, 0].grid(axis='y', linestyle='--', alpha=0.5)

# 2. Grafik Garis Realisasi Kumulatif vs Pagu
axes[0, 1].plot(bulanan['Bulan'], bulanan['Kumulatif'] / 1e9, marker='o', linewidth=2.5, color='#1F4E79', label='Realisasi Kumulatif')
axes[0, 1].axhline(total_pagu / 1e9, color='#D97706', linestyle='--', linewidth=2, label='Total Pagu (Rp11,46 M)')
axes[0, 1].set_title('Realisasi Kumulatif Dibandingkan Pagu Anggaran', fontweight='bold')
axes[0, 1].set_ylabel('Miliar Rupiah')
axes[0, 1].tick_params(axis='x', rotation=35)
axes[0, 1].legend()
axes[0, 1].grid(True, linestyle='--', alpha=0.5)

# 3. Histogram Realisasi Positif per Rekening-Bulan
positif = df[df['Jumlah SPJ (LS + UP/GU/TU)'] > 0]['Jumlah SPJ (LS + UP/GU/TU)'] / 1e6
axes[1, 0].hist(positif, bins=35, color='#2E75B6', edgecolor='white')
axes[1, 0].set_title('Distribusi Realisasi Positif per Baris Rekening-Bulan', fontweight='bold')
axes[1, 0].set_xlabel('Juta Rupiah (Nilai > 0)')
axes[1, 0].set_ylabel('Frekuensi')
axes[1, 0].grid(axis='y', linestyle='--', alpha=0.5)

# 4. Perbandingan Pagu dan Realisasi per Program
prog_pagu = df[kunci_rekening + ['Jumlah Anggaran']].drop_duplicates().groupby('Program')['Jumlah Anggaran'].sum()
prog_real = df.groupby('Program')['Jumlah SPJ (LS + UP/GU/TU)'].sum()
prog_df = pd.DataFrame({'Pagu': prog_pagu, 'Realisasi': prog_real}).sort_values('Pagu')
y_pos = np.arange(len(prog_df))
label_pendek = [p.replace('PROGRAM ', '')[:32] + ('...' if len(p) > 40 else '') for p in prog_df.index]

axes[1, 1].barh(y_pos - 0.18, prog_df['Pagu'] / 1e9, height=0.35, color='#94A3B8', label='Pagu Anggaran')
axes[1, 1].barh(y_pos + 0.18, prog_df['Realisasi'] / 1e9, height=0.35, color='#D97706', label='Realisasi')
axes[1, 1].set_yticks(y_pos)
axes[1, 1].set_yticklabels(label_pendek, fontsize=8.5)
axes[1, 1].set_xlabel('Miliar Rupiah')
axes[1, 1].set_title('Pagu vs Realisasi menurut Program', fontweight='bold')
axes[1, 1].legend()
axes[1, 1].grid(axis='x', linestyle='--', alpha=0.5)

plt.tight_layout()
plt.show()
```

```python
# ==============================================================================
# SKRIP LENGKAP EDA LPJ FUNGSIONAL SIPD - KECAMATAN PURWOKERTO UTARA (JAN-OKT 2026)
# ==============================================================================
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
from scipy.stats import skew, kurtosis

# 1. Memuat Dataset
file_path = '/content/drive/MyDrive/Tugas_AMIKOM/FIK Olah SIPD - LPJ Fungsional - Januari s d Oktober 2026.xlsx'
df = pd.read_excel(file_path, sheet_name='LPJ Administratif 2026')

# 2. Normalisasi Label Bulan
df['Bulan'] = df['Bulan'].astype(str).str.strip()
urutan_bulan = ['Januari', 'Februari', 'Maret', 'April', 'Mei',
                'Juni', 'Juli', 'Agustus', 'September', 'Oktober']
df['Bulan'] = pd.Categorical(df['Bulan'], categories=urutan_bulan, ordered=True)

# 3. Konversi Format Mata Uang Rupiah ke Float
def rupiah_to_float(x):
    if pd.isna(x):
        return np.nan
    if isinstance(x, (int, float, np.number)):
        return float(x)
    x = str(x).replace('Rp', '').replace('.', '').replace(',', '.').strip()
    return float(x)

kolom_nominal = ['Jumlah Anggaran', 'SPJ - LS Gaji', 'SPJ - LS Barang & Jasa',
                 'SPJ - UP/GU/TU', 'Jumlah SPJ (LS + UP/GU/TU)']
for col in kolom_nominal:
    df[col] = df[col].map(rupiah_to_float)

# 4. Rekonsiliasi Komponen SPJ
df['cek_selisih'] = (df['Jumlah SPJ (LS + UP/GU/TU)']
                     - df['SPJ - LS Gaji']
                     - df['SPJ - LS Barang & Jasa']
                     - df['SPJ - UP/GU/TU'])
assert df['cek_selisih'].abs().max() < 1e-6, "Terdapat selisih komponen SPJ!"

# 5. Perhitungan Pagu Unik & Agregasi Bulanan
kunci = ['Sub Unit', 'Program', 'Kegiatan', 'Sub Kegiatan', 'Kode Rekening', 'Uraian']
pagu_total = df[kunci + ['Jumlah Anggaran']].drop_duplicates()['Jumlah Anggaran'].sum()

bulanan = df.groupby('Bulan', observed=True).agg(
    realisasi=('Jumlah SPJ (LS + UP/GU/TU)', 'sum')
).reset_index()
bulanan['kontribusi_pagu_%'] = bulanan['realisasi'] / pagu_total * 100
bulanan['realisasi_kumulatif'] = bulanan['realisasi'].cumsum()
bulanan['persen_serapan'] = bulanan['realisasi_kumulatif'] / pagu_total * 100

print(f"Total Pagu Unik   : Rp{pagu_total:,.2f}")
print(f"Total Realisasi   : Rp{bulanan['realisasi'].sum():,.2f}")
print(f"Persentase Serapan: {bulanan['realisasi'].sum() / pagu_total * 100:.2f}%")
print("\nRingkasan Bulanan:")
print(bulanan.to_string(index=False))

# 6. Ringkasan Mekanisme SPJ
print("\nKomposisi Mekanisme SPJ:")
total_real = df['Jumlah SPJ (LS + UP/GU/TU)'].sum()
for m in ['SPJ - LS Gaji', 'SPJ - LS Barang & Jasa', 'SPJ - UP/GU/TU']:
    val = df[m].sum()
    print(f"- {m}: Rp{val:,.2f} ({val/total_real*100:.2f}%)")
```
